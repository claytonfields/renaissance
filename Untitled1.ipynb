```python
import pandas as pd
import transformers
from transformers import AutoModel, AutoConfig
```

## Pretrained

```python
model = AutoModel.from_pretrained("google/electra-small-discriminator")
model
```

```python
model.config.hidden_size
```

```python
model.embeddings.word_embeddings.weight
```

## Random Init

```python
rand_config = AutoConfig.from_pretrained("google/electra-small-discriminator")
rand_config
```

```python
rand_model = AutoModel.from_config(rand_config)
rand_model
```

```python
rand_model.embeddings.word_embeddings.weight
```

## Custom Dimensions

```python
custom_kwargs = {
    'embedding_size' : 256,
    'hidden_size' : 400
}
custom_config = AutoConfig.from_pretrained("google/electra-small-discriminator", **custom_kwargs)
custom_config
```

```python
custom_model = AutoModel.from_config(custom_config)
custom_model
```
